```python
import pandas as pd
```

```python
df_1=pd.DataFrame({
    "Name":["Ali","Sara","Ahmed"],
    "Roll_No":[101,102,103],
    "Class":[10,11,12]
    
})

df_2=pd.DataFrame({
    "Roll_No":[101,102,103],
    "City":["Kar","Isl","Lah"]
})

type(df_1)
```

```python
df_1_merge_1=pd.merge(df_1,df_2,on="Roll_No",how="inner")
print(df_1_merge_1)
```

```python
df_left=pd.merge(df_1,df_2,on="Roll_No",how="left")
print(df_left)
```

```python
df_right=pd.merge(df_1,df_2,on="Roll_No",how="right")
print(df_right)
```

```python
df_concat=pd.concat(df_1,df_2])
df_concat
```
