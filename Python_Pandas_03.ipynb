```python
data_01={
    "Name":["Ali","Ahmed","Ayesha","Farhan"],
    "Age":[34,45,28,8],
    "Salary":[45000,25000,19000,67000],
    
}
data_02=pd.DataFrame(data_01)
data_02
```

```python
# using loc function

data=data_02.loc[0:1,["Name","Age"]]
data
```

```python
# using iloc function

data=data_02.iloc[0:3,0:4]
data
```

```python
# Boolean Filtering
data=data_02[data_02["Salary"]>20000]
data
```

```python
# Safe Assignment Using loc

data=data_02.loc[data_02["Age"]<20,"Salary"]=50000
data
```
