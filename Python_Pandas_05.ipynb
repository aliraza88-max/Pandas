```python
import pandas as pd
```

```python
data=pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={
    "Category":"category"
    }
)
data.info()
```

```python
# convert profit into numeric

data["Profit"]=pd.to_numeric(data["Profit"])
data.info()
```

```python
# convert region to category
data["Region"]=data["Region"].astype("category")
data.info()
print(data["Region"].memory_usage(deep=True))
```
